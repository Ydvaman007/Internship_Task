```python
eda_df['service fee'] = (
    eda_df['service fee']
    .str.replace('$', '', regex=False)
    .str.strip()
    .astype(float)
)
```

```python
negative_count = (eda_df['minimum nights'] < 0).sum()

print("Negative minimum nights:", negative_count)
```

```python
eda_df = eda_df[eda_df['minimum nights'] >= 0].copy()
```

```python
print("Rows after removing invalid values:", len(eda_df))
print("Negative values remaining:",
      (eda_df['minimum nights'] < 0).sum())
```

```python
eda_df['neighbourhood group'] = (
    eda_df['neighbourhood group']
    .replace({'brookln': 'Brooklyn'})
)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8, 5))

sns.countplot(
    data=eda_df,
    x='room type'
)

plt.title('Distribution of Airbnb Listings by Room Type')
plt.xlabel('Room Type')
plt.ylabel('Number of Listings')
plt.xticks(rotation=15)

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.countplot(
    data=eda_df,
    x='neighbourhood group'
)

plt.title('Distribution of Airbnb Listings by Neighbourhood Group')
plt.xlabel('Neighbourhood Group')
plt.ylabel('Number of Listings')
plt.xticks(rotation=15)

plt.show()
```

```python
plt.figure(figsize=(10, 5))

sns.histplot(
    data=eda_df,
    x='price',
    bins=30,
    kde=True
)

plt.title('Distribution of Airbnb Prices')
plt.xlabel('Price')
plt.ylabel('Number of Listings')

plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.boxplot(
    data=eda_df,
    x='room type',
    y='price'
)

plt.title('Price Distribution by Room Type')
plt.xlabel('Room Type')
plt.ylabel('Price')
plt.xticks(rotation=15)

plt.show()
```

```python
avg_price_room = (
    eda_df.groupby('room type')['price']
    .mean()
    .sort_values(ascending=False)
)

plt.figure(figsize=(9, 5))

avg_price_room.plot(kind='bar')

plt.title('Average Airbnb Price by Room Type')
plt.xlabel('Room Type')
plt.ylabel('Average Price')
plt.xticks(rotation=15)

plt.show()
```

```python
avg_price_neighbourhood = (
    eda_df.groupby('neighbourhood group')['price']
    .mean()
    .sort_values(ascending=False)
)

plt.figure(figsize=(9, 5))

avg_price_neighbourhood.plot(kind='bar')

plt.title('Average Airbnb Price by Neighbourhood Group')
plt.xlabel('Neighbourhood Group')
plt.ylabel('Average Price')
plt.xticks(rotation=15)

plt.show()
```

```python
plt.figure(figsize=(10, 5))

sns.histplot(
    data=eda_df,
    x='minimum nights',
    bins=30
)

plt.title('Distribution of Minimum Nights')
plt.xlabel('Minimum Nights')
plt.ylabel('Number of Listings')

plt.show()
```

```python
plt.figure(figsize=(10, 5))

sns.histplot(
    data=eda_df,
    x='number of reviews',
    bins=30,
    kde=True
)

plt.title('Distribution of Number of Reviews')
plt.xlabel('Number of Reviews')
plt.ylabel('Number of Listings')

plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=eda_df,
    x='price',
    y='number of reviews',
    alpha=0.4
)

plt.title('Price vs Number of Reviews')
plt.xlabel('Price')
plt.ylabel('Number of Reviews')

plt.show()
```

```python
numeric_df = eda_df.select_dtypes(include='number')

plt.figure(figsize=(12, 8))

sns.heatmap(
    numeric_df.corr(),
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title('Correlation Matrix of Numerical Variables')

plt.show()
```

```python
neighbourhood_summary = eda_df.groupby('neighbourhood group').agg(
    listings=('NAME', 'count'),
    average_price=('price', 'mean'),
    median_price=('price', 'median'),
    average_reviews=('number of reviews', 'mean'),
    average_minimum_nights=('minimum nights', 'mean')
).sort_values('average_price', ascending=False)

neighbourhood_summary.round(2)
```

```python
room_neighbourhood = pd.pivot_table(
    eda_df,
    values='price',
    index='neighbourhood group',
    columns='room type',
    aggfunc='mean'
)

room_neighbourhood.round(2)
```
